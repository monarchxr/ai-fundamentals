```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0,2.0])
y_train = np.array([300.0, 500.0])
print(f"x_train={x_train}")
print(f"y_train={y_train}")
```

```python
print(f"x_train.shape: {x_train.shape}") # .shape returns a tuple, with an entry for each dimension

# m is the number of training examples
m = x_train.shape[0]
print(f"Number of training examples, m = {m}")
```

```python
# lets make ith training example
i = 1

x_i = x_train[i]
y_i = y_train[i]

print(f"(x^{i}, y^{i}) = ({x_i}, {y_i})")
```

```python
# plotting

plt.scatter(x_train, y_train, marker='x', c='r') # marker you get what it is, c is for colour
plt.title("House prices")
plt.ylabel("Price/1000$")
plt.xlabel("Size/1000 sqft")

plt.show()
```

```python
# now deal with model's parameters
# w and b

# w = weight
# b = bias

w = 200
b = 100
print(f"w = {w}")
print(f"b = {b}")
```

```python
# lets make the prediction model function
def compute_model_output(x, w, b):
    m = x.shape[0]

    f_wb = np.zeros(m) #creates an array filled entirely with 0s
    for i in range(m):
        f_wb[i] = w*x[i] + b

    return f_wb
```

```python
temp_fwb = compute_model_output(x_train, w,b)

plt.plot(x_train, temp_fwb, c='b', label='Our prediction')
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual value')
plt.title("House prices")
plt.ylabel("Prices/1000$")
plt.xlabel("Size/1000 sqft")

plt.legend()
plt.show()
```

```python
# lets now predict the price of a house of size 1200 sqft
# that would become 1.2, in our units

w = 200
b = 100
x_i = 1.2
y_i = w*x_i+b

print(f"${y_i: .0f} thousand dollars")
```
