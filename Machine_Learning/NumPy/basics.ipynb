Now let's work on the Dot Product

```python
def manual(a,b):
    x = 0

    for i in range(a.shape[0]):
        x += a[i]*b[i]
    
    return x
```

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

print(np.dot(a,b))
print(manual(a,b))
```

Now let's perform a comparison

```python
np.random.seed(1)

a = np.random.rand(100000)
b = np.random.rand(100000)

tick = time.time()
c = np.dot(a,b)
tock = time.time()

print(f"Vectorized version duration = {1000*(tock-tick)} milliseconds")

tick = time.time()
c = manual(a,b)
tock = time.time()

print(f"Manual loop version duration = {1000*(tock-tick)} milliseconds")

del(a)
del(b)
```
